```python
import tensorflow as tf

print("TensorFlow:", tf.__version__)
print("Devices:", tf.config.list_physical_devices())
```

```python
import pandas as pd

df = pd.read_csv("gestures_augmented.csv", low_memory=False)
```

```python
print(df.shape)
```

```python
x = df.drop(columns=["label"])
y = df["label"]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.4,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)
```

```python
print("df:", df.shape)
print("x_train:", x_train.shape)
print("x_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
print("Classes:", len(label_encoder.classes_))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

print("Training shape:", x_train_scaled.shape)
print("Testing shape:", x_test_scaled.shape)
print("Mean:", round(x_train_scaled.mean(), 5))
print("Std:", round(x_train_scaled.std(), 5))
```

```python
NUM_TIMESTEPS = 100
NUM_FEATURES = 6

x_train_lstm = x_train_scaled.reshape(
    x_train_scaled.shape[0],
    NUM_TIMESTEPS,
    NUM_FEATURES
)

x_test_lstm = x_test_scaled.reshape(
    x_test_scaled.shape[0],
    NUM_TIMESTEPS,
    NUM_FEATURES
)

print("Training:", x_train_lstm.shape)
print("Testing :", x_test_lstm.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, BatchNormalization, Input

tf.random.set_seed(42)

num_classes = len(label_encoder.classes_)

lstm_model = Sequential([
    Input(shape=(100, 6)),

    LSTM(64, return_sequences=True),
    Dropout(0.25),

    LSTM(32),
    BatchNormalization(),
    Dropout(0.25),

    Dense(64, activation="relu"),
    Dropout(0.2),

    Dense(num_classes, activation="softmax")
])

lstm_model.compile(
    optimizer="adam",
    loss="sparse_cateogorical_crossentropy",
    metrics=["accuracy"]
)

lstm_model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=8,
    restore_best_weights=True,
    verbose=1
)
```

```python
history = lstm_model.fit(
    x_train_lstm,
    y_train_encoded,

    validation_split=0.2,

    epochs=50,
    batch_size=32,

    callbacks=[early_stop],

    verbose=1
)
```

```python
lstm_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
from tensorflow.keras.losses import SparseCategoricalCrossentropy

lstm_model.compile(
    optimizer="adam",
    loss=SparseCategoricalCrossentropy(),
    metrics=["accuracy"]
)
```

```python
import tensorflow as tf
import keras

print(tf.__version__)
print(keras.__version__)
```

```python
history = lstm_model.fit(
    x_train_lstm,
    y_train_encoded,
    validation_split=0.2,
    epochs=50,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)
```

```python
test_loss, test_accuracy = lstm_model.evaluate(
    x_test_lstm,
    y_test_encoded,
    verbose=1
)

print(f"Test accuracy: {test_accuracy:.4f}")
print(f"Test loss: {test_loss:.4f}")
```

```python
import numpy as np

y_prob_lstm = lstm_model.predict(x_test_lstm, verbose=1)
y_pred_lstm = np.argmax(y_prob_lstm, axis=1)

print("First 20 predicted classes:")
print(label_encoder.inverse_transform(y_pred_lstm[:20]))
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test_encoded,
        y_pred_lstm,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

fig, ax = plt.subplots(figsize=(18, 18))

ConfusionMatrixDisplay.from_predictions(
    y_test_encoded,
    y_pred_lstm,
    display_labels=label_encoder.classes_,
    normalize="true",
    xticks_rotation=90,
    values_format=".2f",
    ax=ax
)

plt.title("Normalized Confusion Matrix - LSTM")
plt.tight_layout()
plt.show()
```

```python
import joblib

lstm_model.save("lstm_air_writing.keras")
joblib.dump(scaler, "lstm_scaler.pkl")
joblib.dump(label_encoder, "label_encoder.pkl")

print("Saved LSTM model, scaler, and label encoder.")
```

```python
# Comparing my live recordings against each Kaggle class
# using per-channel mean and standard deviation

results = []

live_mean = live_ch.mean(axis=(0, 1))
live_std = live_ch.std(axis=(0, 1))

for label in sorted(df["label"].unique()):

    class_data = (
        df[df["label"] == label]
        .drop(columns=["label"])
        .to_numpy(dtype=float)
        .reshape(-1, 100, 6)
    )

    class_mean = class_data.mean(axis=(0, 1))
    class_std = class_data.std(axis=(0, 1))

    mean_distance = np.linalg.norm(
        class_mean - live_mean
    )

    std_distance = np.linalg.norm(
        class_std - live_std
    )

    total_distance = mean_distance + std_distance

    results.append(
        (label, mean_distance, std_distance, total_distance)
    )

results.sort(key=lambda x: x[3])

print(
    f"{'Label':<8}"
    f"{'Mean Dist':>12}"
    f"{'Std Dist':>12}"
    f"{'Total':>12}"
)

print("-" * 44)

for label, md, sd, total in results[:10]:
    print(
        f"{label:<8}"
        f"{md:>12.4f}"
        f"{sd:>12.4f}"
        f"{total:>12.4f}"
    )
```

```python
# Examine where the largest movements occur in each recording

names = ["Ax", "Ay", "Az", "Gx", "Gy", "Gz"]

for rec_num, rec in enumerate(live_ch, start=1):

    print(f"\nRecording {rec_num}")
    print("-" * 55)

    for j, name in enumerate(names):

        values = rec[:, j]

        # Largest absolute change between consecutive samples
        diffs = np.abs(np.diff(values))
        max_change_idx = np.argmax(diffs)

        print(
            f"{name}: "
            f"min={values.min():8.4f}, "
            f"max={values.max():8.4f}, "
            f"largest Δ={diffs[max_change_idx]:8.4f}"
            f"at samples {max_change_idx}->{max_change_idx+1}"
        )
```

```python
# Re-running the class comparison using ONLY clean recordings 2 and 3

clean_live = live_ch[1:3]

clean_mean = clean_live.mean(axis=(0, 1))
clean_std = clean_live.std(axis=(0, 1))

clean_results = []

for label in sorted(df["label"].unique()):

    class_data = (
        df[df["label"] == label]
        .drop(columns=["label"])
        .to_numpy(dtype=float)
        .reshape(-1, 100, 6)
    )

    class_mean = class_data.mean(axis=(0, 1))
    class_std = class_data.std(axis=(0, 1))

    mean_distance = np.linalg.norm(
        class_mean - clean_mean
    )

    std_distance = np.linalg.norm(
        class_std - clean_std
    )

    total_distance = mean_distance + std_distance

    clean_results.append(
        (label, mean_distance, std_distance, total_distance)
    )

clean_results.sort(key=lambda x: x[3])

print(
    f"{'Rank':<6}"
    f"{'Label':<8}"
    f"{'Mean Dist':>12}"
    f"{'Std Dist':>12}"
    f"{'Total':>12}"
)

print("-" * 50)

for rank, (label, md, sd, total) in enumerate(
    clean_results, start=1
):
    print(
        f"{rank:<6}"
        f"{label:<8}"
        f"{md:>12.4f}"
        f"{sd:>12.4f}"
        f"{total:>12.4f}"
    )

# Specifically report A
for rank, (label, md, sd, total) in enumerate(
    clean_results, start=1
):
    if label == "A":
        print("\nA WITH RECORDING 1 REMOVED")
        print(f"Rank: {rank} / {len(clean_results)}")
        print(f"Mean distance: {md:.4f}")
        print(f"Std distance: {sd:.4f}")
        print(f"Total distance: {total:.4f}")
        break
```

```python
import itertools
import numpy as np

# Use only clean recordings 2 and 3
base_live = live_ch[1:3]

axis_names = ["Ax", "Ay", "Az"]
gyro_names = ["Gx", "Gy", "Gz"]

best_results = []

# Permute accelerometer axes and gyro axes together
# and test sign flips
for perm in itertools.permutations([0, 1, 2]):

    for signs in itertools.product([-1, 1], repeat=3):

        transformed = base_live.copy()

        # Accelerometer
        acc = base_live[:, :, :3][:, :, perm]
        acc = acc * np.array(signs)

        # Gyroscope uses same axis mapping/sign convention
        gyro = base_live[:, :, 3:][:, :, perm]
        gyro = gyro * np.array(signs)

        transformed[:, :, :3] = acc
        transformed[:, :, 3:] = gyro

        live_mean = transformed.mean(axis=(0, 1))
        live_std = transformed.std(axis=(0, 1))

        results_test = []

        for label in sorted(df["label"].unique()):

            class_data = (
                df[df["label"] == label]
                .drop(columns=["label"])
                .to_numpy(dtype=float)
                .reshape(-1, 100, 6)
            )

            class_mean = class_data.mean(axis=(0, 1))
            class_std = class_data.std(axis=(0, 1))

            md = np.linalg.norm(class_mean - live_mean)
            sd = np.linalg.norm(class_std - live_std)
            total = md + sd 

            results_test.append((label, total))

        results_test.sort(key=lambda x: x[1])

        labels_ranked = [x[0] for x in results_test]
        a_rank = labels_ranked.index("A") + 1
        a_distance = [x[1] for x in results_test if x[0] == "A"][0]

        best_results.append(
            (a_rank, a_distance, perm, signs)
        )
best_results.sort(key=lambda x: (x[0], x[1]))

print("Best axis mappings for making live data resemble Kaggle A:\n")

for rank, dist, perm, signs in best_results[:10]:
    print(
        f"A rank={rank:2d}, "
        f"distance={dist:.4f}, "
        f"perm={perm}, "
        f"signs={signs}"
    ) 
```

```python
# Comparing each clean live recording directly against every individual Kaggle A recording using time-series distance

kaggle_A = (
    df[df["label"] == "A"]
    .drop(columns=["label"])
    .to_numpy(dtype=float)
    .reshape(-1, 100, 6)
)

clean_live = live_ch[1:3]

# Best orientation discovered in the previous test
perm = (2, 0, 1)
signs = np.array([1, 1, -1])

transformed_live = clean_live.copy()

transformed_live [:, :, :3] = (
    clean_live[:, :, :3][:, :, perm] * signs
)

print("Kaggle A recordings:", kaggle_A.shape[0])
print()

for live_num, live_rec in enumerate(transformed_live, start=2):

    distances = []

    for i, kaggle_rec in enumerate(kaggle_A):

        # Root mean squared error across the entire 100x6 sequence 
        rmse = np.sqrt(
            np.mean((kaggle_rec - live_rec) ** 2)
        )

        distances.append((i, rmse))

    distances.sort(key=lambda x: x[1])

    print(f"Live Recording {live_num}")
    print("-" * 40)

    for kaggle_idx, distance in distances[:10]:
        print(
            f"Kaggle A #{kaggle_idx}: "
            f"RMSE = {distance:.4f}"
        )

    print()
```

```python
# Measuring normal within-class variation among Kaggle A recordings

pairwise_rmse = []

for i in range(len(kaggle_A)):
    for j in range(i + 1, len(kaggle_A)):

        rmse = np.sqrt(
            np.mean((kaggle_A[i] - kaggle_A[j]) ** 2)
        )

        pairwise_rmse.append(rmse)

pairwise_rmse = np.array(pairwise_rmse)

print("Kaggle A-to-A RMSE statistics")
print("-" * 40)
print(f"Mean RMSE:   {pairwise_rmse.mean():.4f}")
print(f"Median RMSE: {np.median(pairwise_rmse):.4f}")
print(f"Std RMSE:    {pairwise_rmse.std():.4f}")
print(f"Minimum:     {pairwise_rmse.min():.4f}")
print(f"Maximum:     {pairwise_rmse.max():.4f}")

print("\nPercentiles:")
for p in [5, 25, 50, 75, 95]:
    print(
        f"{p:>2}th percentile: "
        f"{np.percentile(pairwise_rmse, p):.4f}"
    )
```

```python
import numpy as np
import pandas as pd

def parse_recording(text):
    values = np.array([
        float(x.strip())
        for x in text.replace("\n", "").split(",")
        if x.strip()
    ])

    if len(values) != 600:
        raise ValueError(
            f"Expected 600 values, but got {len(values)}"
        )

    return values
```

```python
A_recordings = [parse_recording(x) for x in A_texts]
B_recordings = [parse_recording(x) for x in B_texts]
C_recordings = [parse_recording(x) for x in C_texts]

print("A recordings:", len(A_recordings))
print("B recordings:", len(B_recordings))
print("C recordings:", len(C_recordings))

print("Length of first A:", len(A_recordings[0]))
print("Length of first B:", len(B_recordings[0]))
print("Length of first C:", len(C_recordings[0]))
```

```python
import numpy as np

X = np.vstack(
    A_recordings +
    B_recordings +
    C_recordings
)

y = np.array(
    ["A"] * len(A_recordings) +
    ["B"] * len(B_recordings) +
    ["C"] * len(C_recordings)
)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Labels:", y)
```

```python
from sklearn.model_selection import LeaveOneOut
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
import numpy as np

loo = LeaveOneOut()

loo_actual = []
loo_predicted = []

for train_index, test_index in loo.split(X):

    X_train_loo = X[train_index]
    X_test_loo = X[test_index]

    y_train_loo = y[train_index]
    y_test_loo = y[test_index]

    model = RandomForestClassifier(
        n_estimators = 300,
        random_state = 42
    )

    model.fit(X_train_loo, y_train_loo)

    prediction = model.predict(X_test_loo)[0]

    loo_actual.append(y_test_loo[0])
    loo_predicted.append(prediction)

loo_actual = np.array(loo_actual)
loo_predicted = np.array(loo_predicted)

print("Leave-One-Recording-Out Results")
print("-" * 45)

for i, (actual, predicted) in enumerate(
    zip(loo_actual, loo_predicted),
    start = 1
):
    result = "CORRECT" if actual == predicted else "WRONG"

    print(
        f"Recording {i:02d}: "
        f"Actual = {actual}, "
        f"Predicted = {predicted} [{result}]" 
    )

loo_accuracy = accuracy_score(
    loo_actual,
    loo_predicted
)

print("\nLOO Accuracy:", loo_accuracy)
print("LOO Accuracy percentage:", loo_accuracy * 100, "%")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    loo_actual,
    loo_predicted,
    labels=["A", "B", "C"],
    display_labels = ["A", "B", "C"],
    normalize = "true"
)

plt.title(
    "MPU6050 Air-Writing - "
    "Leave-One-Recording-Out Random Forest"
)

plt.show()
```

```python
from sklearn.model_selection import train_test_split
import numpy as np 

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42, 
    stratify = y
)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)

print("\nTraining labels:")
print(np.unique(y_train, return_counts = True))

print("\nTesting labels:")
print(np.unique(y_test, return_counts = True))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

rf_model = RandomForestClassifier(
    n_estimators = 500,
    random_state = 42,
    max_depth = None
)

rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

print("Actual:  ", y_test)
print("Predicted:", y_pred)

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        labels = ["A", "B", "C"],
        zero_division = 0
    )
)
```

```python
print("Raw A recordings:", len(A_texts))
print("Raw B recordings:", len(B_texts))
print("Raw C recordings:", len(C_texts))

for name, texts in [("A", A_texts), ("B", B_texts), ("C", C_texts)]:
    lengths = []

    for i, text in enumerate(texts, start = 1):
        values = np.fromstring(text, sep = ",")
        lengths.append(len(values))

        if len(values) != 600:
            print(f"{name} sample {i}: {len(values)} values <-- CHECK THIS")

    print(f"{name}: {len(texts)} samples")
    print(f"{name} value counts:", lengths)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
import numpy as np

cv = StratifiedKFold(
    n_splits = 5, 
    shuffle = True,
    random_state = 42
)

rf_cv = RandomForestClassifier(
    n_estimators = 500,
    random_state = 42,
    max_depth = None
)

scores = cross_val_score(
    rf_cv,
    X,
    y,
    cv = cv,
    scoring = "accuracy"
)

print("5-Fold Cross-Validation Results")
print("-" * 40)

for i, score in enumerate(scores, start = 1):
    print(f"Fold {i}: {score * 100:.2f}%")

print("\nMean accuracy:", scores.mean())
print("Mean accuracy percentage:", scores.mean() * 100, "%")
print("Standard deviation:", scores.std())
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

cv_predictions = cross_val_predict(
    rf_cv,
    X,
    y,
    cv = cv
)

cv_accuracy = accuracy_score(y, cv_predictions)

print("Cross-validated accuracy:", cv_accuracy)
print("Cross-validated accuracy percentage:", 
      cv_accuracy * 100, "%")

print("\nClassification Report:")
print(
    classification_report(
        y,
        cv_predictions,
        labels = ["A", "B", "C"],
        zero_division = 0
    )
)

ConfusionMatrixDisplay.from_predictions(
    y,
    cv_predictions,
    labels = ["A", "B", "C"],
    display_labels = ["A", "B", "C"],
    normalize = "true"
)

plt.title("MPU6050 Air-Writing - 5-Fold Cross-Validated Random Forest")
plt.show()
```

```python
channel_names = [
    "Accel X",
    "Accel Y",
    "Accel Z",
    "Gyro X",
    "Gyro Y",
    "Gyro Z"
]

def plot_recording(recording, title):
    data = recording.reshape(100, 6)

    time = np.arange(100) * 0.04   # 40 ms = 0.04 seconds

    plt.figure(figsize=(12, 6))

    for channel in range(6):
        plt.plot(
            time,
            data[:, channel],
            label=channel_names[channel]
        )

    plt.xlabel("Time (seconds)")
    plt.ylabel("Sensor value")
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()


plot_recording(
    X[b19_index],
    "Misclassified B Recording 19 — Predicted C"
)

plot_recording(
    X[c01_index],
    "Misclassified C Recording 01 — Predicted A"
)
```

```python
# Parse the 30 completely unseen recordings
A_new = np.array([parse_recording(text) for text in A_new_texts])
B_new = np.array([parse_recording(text) for text in B_new_texts])
C_new = np.array([parse_recording(text) for text in C_new_texts])

print("A:", A_new.shape)
print("B:", B_new.shape)
print("C:", C_new.shape)
```

```python
print("A strings:", len(A_new_texts))
print("B strings:", len(B_new_texts))
print("C strings:", len(C_new_texts))

for name, texts in [
    ("A", A_new_texts),
    ("B", B_new_texts),
    ("C", C_new_texts)
]:
    print(f"\n{name}:")

    for i, text in enumerate(texts, start = 1):
        pieces = [
            x.strip()
            for x in text.replace("\n", "").split(",")
            if x.strip()
        ]

        print(
            f"{name}{i:02d}: "
            f"{len(pieces)} comma-seperated values"
        )
```

```python
X_unseen = np.vstack([
    A_new,
    B_new,
    C_new
])

y_unseen = np.array(
    ["A"] * 10 +
    ["B"] * 10 +
    ["C"] * 10
)

print("Unseen X shape:", X_unseen.shape)
print("Unseen y shape:", y_unseen.shape)
print("Labels:", np.unique(y_unseen, return_counts = True))
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)

import matplotlib.pyplot as plt

unseen_predictions = final_rf.predict(X_unseen)

print("Actual:")
print(y_unseen)

print("\nPredicted:")
print(unseen_predictions)

unseen_accuracy = accuracy_score(
    y_unseen,
    unseen_predictions
)

print("\nUNSEEN TEST ACCURACY")
print("--------------------")
print("Accuracy:", unseen_accuracy)
print("Accuracy percentage:", unseen_accuracy * 100, "%")

print("\nClassification Report:")
print(
    classification_report(
        y_unseen,
        unseen_predictions,
        labels = ["A", "B", "C"], 
        zero_division = 0
    )
)

ConfusionMatrixDisplay.from_predictions(
    y_unseen,
    unseen_predictions,
    labels = ["A", "B", "C"],
    display_labels = ["A", "B", "C"],
    normalize = "true"
)

plt.title(
    "MPU6050 Air-Writing - Completely Unseen Test Data"
)
plt.show()
```

```python
print("Misclassified unseen recordings")
print("-" * 45)

for i, (actual, predicted) in enumerate(
    zip(y_unseen, unseen_predictions)
):

    if i < 10:
        letter = "A"
        recording = i + 1

    elif i < 20:
        letter = "B"
        recording = i - 10 + 1

    else:
        letter = "C"
        recording = i - 20 + 1

    if actual != predicted:
        print(
            f"{letter} Recording {recording:02d}: "
            f"Actual = {actual}, "
            f"Predicted = {predicted}"
        )
```

```python
unseen_probabilities = final_rf.predict_proba(X_unseen)

print("Prediction confidence on unseen recordings")
print("-" * 55)

for i, (actual, predicted, probs) in enumerate(
    zip(y_unseen, unseen_predictions, unseen_probabilities)
):
    if i < 10:
        recording = i + 1
    elif i < 20:
        recording = i - 10 + 1
    else:
        recording = i - 20 + 1

    predicted_index = list(final_rf.classes_).index(predicted)
    confidence = probs[predicted_index]

    print(
        f"{actual} Recording {recording:02d}: "
        f"Predicted = {predicted}, "
        f"Confidence = {confidence * 100:.2f}%"
    )
```

```python
max_confidences = unseen_probabilities.max(axis = 1)

print("\nConfidence summary")
print("-" * 30)
print("Minimum confidence:", max_confidences.min() * 100, "%")
print("Mean confidence:", max_confidences.mean() * 100, "%")
print("Median confidence:", np.median(max_confidences) * 100, "%")

lowest_indices = np.argsort(max_confidences)[:5]

print("\nFive least-confident predictions:")
for idx in lowest_indices:
    print(
        f"Sample {idx + 1:02d}: "
        f"Actual = {y_unseen[idx]}, "
        f"Predicted = {unseen_predictions[idx]}, "
        f"Confidence = {max_confidences[idx] * 100:.2f}%"
    )
```

```python
print("Detailed probability analysis")
print("-" * 65)

for idx in lowest_indices:

    if idx < 10:
        recording_name = f"A Recording {idx + 1:02d}"
    elif idx < 20:
        recording_name = f"B Recording {idx - 10 + 1:02d}"
    else:
        recording_name = f"C Recording {idx - 20 + 1:02d}"

    probs = unseen_probabilities[idx]

    print(f"\n{recording_name}")
    print(f"Actual class:    {y_unseen[idx]}")
    print(f"Predicted class: {unseen_predictions[idx]}")

    for class_name, probability in zip(final_rf.classes_, probs):
        print(
            f"  P({class_name}) = "
            f"{probability * 100:.2f}%"
        )
```

```python
print("\nConfidence by class")
print("-" * 40)

for letter in final_rf.classes_:

    mask = y_unseen == letter
    class_confidences = max_confidences[mask]

    print(
        f"{letter}: "
        f"mean = {class_confidences.mean() * 100:.2f}%, "
        f"min = {class_confidences.min() * 100:.2f}%, "
        f"max = {class_confidences.max() * 100:.2f}%"
    )
```

```python
# Robustness test: Adding small amounts of noise to unseen recordings

rng = np.random.default_rng(42)

noise_levels = [0.01, 0.02, 0.05, 0.10]

print("Noise robustness test")
print("-" * 55)

for noise_level in noise_levels:

    # Scaling noise relative to the variation of each feature
    feature_std = X_unseen.std(axis = 0)

    noise = rng.normal(
        loc = 0,
        scale = noise_level * feature_std,
        size = X_unseen.shape
    )

    X_noisy = X_unseen + noise

    noisy_predictions = final_rf.predict(X_noisy)

    noisy_accuracy = accuracy_score(
        y_unseen,
        noisy_predictions
    )

    changed_predictions = np.sum(
        noisy_predictions != unseen_predictions
    )

    print(
        f"Noisy level {noise_level * 100:>5.1f}% | "
        f"Accuracy = {noisy_accuracy * 100:6.2f}% | "
        f"Changed predictions = {changed_predictions}/30"
    )

    
```
